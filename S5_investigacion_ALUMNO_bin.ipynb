# FASE 4 — Detección de manipulación (pensamiento crítico avanzado)

Un analista experto no solo crea buenos gráficos: detecta cuándo alguien intenta engañarlo. Aquí generarás gráficos ENGAÑOSOS a propósito, para aprender a reconocerlos.

## Paso 8 — Fabrica un gráfico engañoso


```python
# El MISMO dato (rating por categoría) presentado de forma engañosa vs honesta
rating_cat = df.groupby("category")["rating"].mean().sort_values(ascending=False).head(6)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
# Version que EXAGERA (eje truncado)
axes[0].bar(range(len(rating_cat)), rating_cat.values, color="#e2231a")
axes[0].set_ylim(3.9, 4.05)
axes[0].set_title("Versión A")
axes[0].set_xticks(range(len(rating_cat)))
axes[0].set_xticklabels(rating_cat.index, rotation=45, ha="right", fontsize=7)
# Version honesta
axes[1].bar(range(len(rating_cat)), rating_cat.values, color="#1e2a6b")
axes[1].set_ylim(0, 5)
axes[1].set_title("Versión B")
axes[1].set_xticks(range(len(rating_cat)))
axes[1].set_xticklabels(rating_cat.index, rotation=45, ha="right", fontsize=7)
plt.tight_layout(); plt.show()
```

> ✍️ **26.** Las dos versiones muestran los MISMOS datos. ¿Cuál exagera las diferencias y cómo lo logra? Da el rango exacto del eje Y de cada una.

# FASE 5 — Validación cruzada e integración

Aquí conectas la visualización con lo que aprendiste en las semanas anteriores (estadística, correlación). Una buena visualización debe ser consistente con las cifras.

## Paso 9 — ¿El gráfico dice lo mismo que la estadística?


```python
from scipy import stats

# Comparamos rating entre productos caros y baratos: gráfico + prueba estadística
df["es_caro"] = df["actual_price"] >= df["actual_price"].quantile(0.75)

fig = px.box(df, x="es_caro", y="rating", title="Rating: productos caros vs no caros")
fig.show()

caros = df[df["es_caro"]]["rating"]
baratos = df[~df["es_caro"]]["rating"]
t, p = stats.ttest_ind(caros, baratos, equal_var=False)
print(f"Rating medio caros: {caros.mean():.3f}  |  baratos: {baratos.mean():.3f}")
print(f"Prueba t: p-valor = {p:.4f}")
```

> ✍️ **31.** El boxplot, ¿sugiere una diferencia grande o pequeña de rating entre caros y baratos? ¿El p-valor (¿significativo o no?) confirma lo que ves en el gráfico?

> ✍️ **32.** ¿Por qué es una buena práctica acompañar un gráfico con una prueba estadística? ¿Un gráfico puede "sugerir" una diferencia que en realidad no es significativa?

> ✍️ **33.** Aquí integras visualización (Semana 5) con inferencia (Semana 3). Explica cómo se complementan: ¿qué aporta el gráfico que el p-valor no, y viceversa?


Respuesta 31: El boxplot sugiere una diferencia muy pequeña entre caros y baratos. El p-valor es 0.4341, mucho mayor que 0.05, así que la diferencia no es estadísticamente significativa.

Respuesta 33: El gráfico nos dice la forma y magnitud de la diferencia; el p-valor nos dice si esa diferencia es confiable o no. Se complementan: el primero ayuda a visualizar, el segundo ayuda a inferir.

Respuesta 32: Es buena práctica acompañar el gráfico con una prueba estadística porque un gráfico puede sugerir una diferencia visual, pero no prueba que sea real. Las pruebas cuantifican si esa diferencia es probablemente producto del azar.

## Paso 10 — Cuidado con las conclusiones apresuradas


```python
# Un patrón visual puede ser un artefacto. Veamos rating_count en escala normal vs log
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
axes[0].hist(df["rating_count"], bins=50, color="#1e2a6b")
axes[0].set_title("rating_count (escala normal)")
axes[1].hist(np.log1p(df["rating_count"]), bins=50, color="#e2231a")
axes[1].set_title("rating_count (escala log)")
plt.tight_layout(); plt.show()
```

## Proyecto final: El precio no predice la calidad en el catálogo de Amazon

### Mensaje central
El precio y el descuento no son buenos indicadores de la calidad percibida por los clientes. En este catálogo, la historia real se explica mejor por la diversidad de categorías, la demanda y el comportamiento del rating, no por el hecho de que un producto sea más caro.

### Introducción
El dataset de Amazon reúne productos de distintas categorías, con variables como precio real, descuento, rating y número de reseñas. La pregunta principal fue verificar si un producto más caro o más descontado tiene mejor valoración por parte de los compradores.

### Visualizaciones principales

1. Scatter de precio real vs. rating
   - Tipo: dispersión.
   - ¿Por qué?: permite examinar si existe una relación lineal entre el precio y la satisfacción del cliente.
   - Qué muestra: la nube de puntos es muy dispersa y la tendencia es prácticamente plana. La correlación entre precio real y rating es -0.0352, casi nula.

2. Barras del descuento medio por categoría
   - Tipo: barras comparativas.
   - ¿Por qué?: compara la intensidad de descuento entre categorías y ayuda a descubrir si la promoción explica diferencias en la valoración.
   - Qué muestra: existen diferencias notables entre categorías, pero no son suficientes para explicar un patrón claro de mejor o peor rating.

3. Boxplot del rating por categoría
   - Tipo: boxplot.
   - ¿Por qué?: permite ver la dispersión y la mediana de la valoración por grupo.
   - Qué muestra: los ratings son muy homogéneos en la mayoría de las categorías, con diferencias pequeñas y sin una relación directa con el precio.

4. Histograma de rating_count en escala log
   - Tipo: histograma con escala logarítmica.
   - ¿Por qué?: los conteos de reseñas tienen una distribución muy sesgada y esta escala revela mejor la estructura subyacente.
   - Qué muestra: la mayoría de los productos recibe un volumen bajo de reseñas, mientras que unos pocos tienen explosiones de demanda.

5. Scatter de precio promedio vs. rating promedio por categoría
   - Tipo: dispersión por agregación.
   - ¿Por qué?: compara categorías enteras para detectar si las más caras son también las mejor valoradas.
   - Qué muestra: no aparece una relación consistente; algunas categorías caras tienen ratings similares a categorías más económicas.

### Afirmaciones validadas

- Afirmación verdadera: la categoría con más productos es Electronics, con 525 artículos. Esto confirma que la base de datos está dominada por una categoría muy grande y con gran volumen de transacciones.
- Afirmación falsa: “los productos más caros son los mejor calificados”. La evidencia contradice esta idea: la correlación entre actual_price y rating es -0.0352, casi cero, y el producto más caro tiene un rating de 3.9, no un valor top.

### Lo que los datos desmienten
El hallazgo más contraintuitivo es que el precio no predice la calidad. En una primera impresión, pareciera lógico que lo más caro fuese mejor, pero la evidencia demuestra que la diferencia de precio no explica la valoración de los clientes. Además, la correlación entre discount_percentage y rating es solo 0.0844, lo que confirma que el descuento tampoco es un predictor fuerte.
